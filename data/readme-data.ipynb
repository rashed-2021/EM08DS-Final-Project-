```python
# Data Documentation — `ai_trust_survey_synthetic.csv`

## Source
Data were collected via an anonymous online questionnaire (`questionnaire.html`,
Activity 5.1) distributed to a convenience sample of respondents.

## Sample
N = 45 respondents.

## Variables

| Column | Type | Description | Coding |
|---|---|---|---|
| `respondent_id` | string | Anonymous ID | R001–R045 |
| `age_group` | categorical | Self-reported age bracket | 18-24 / 25-34 / 35-44 / 45+ |
| `gender` | categorical | Self-reported gender | Male / Female / Prefer not to say |
| `education_level` | categorical | Highest education attained | Bachelor's / Master's / Doctorate / Other |
| `technical_background` | categorical | Programming/technical background | Yes / No |
| `ai_familiarity_1to5` | ordinal | Self-rated familiarity with AI systems | 1 (low) – 5 (high) |
| `usage_frequency` | categorical | Frequency of AI tool use | Daily / Weekly / Rarely / Never |
| `q1_trust_low_stakes` | ordinal (Likert) | Trust in AI for low-stakes tasks | 1–5 |
| `q2_trust_high_stakes` | ordinal (Likert) | Trust in AI for high-stakes tasks | 1–5 |
| `q3_perceived_transparency` | ordinal (Likert) | Perceived transparency of AI decisions | 1–5 |
| `q4_would_follow_ai_over_own_judgment` | ordinal (Likert) | Willingness to defer to AI | 1–5 |
| `q5_bias_concern` | ordinal (Likert) | Concern about algorithmic bias | 1–5 |
| `q6_ai_more_objective_than_human` | ordinal (Likert) | Perceived AI objectivity vs. humans | 1–5 |
| `trust_composite_score` | continuous | Mean of Q1, Q2, Q3, Q4, Q6, and reverse-coded Q5 | 1.0–5.0 |
| `trust_label` | binary (target) | Median split of composite score | High Trust / Low Trust |

## Ethical handling
No personally identifying information was collected. Participation was
voluntary and anonymous; respondents could withdraw at any point before
submission. See the Ethics Statement in the main research report.
```
